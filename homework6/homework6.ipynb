```python
import numpy as np

arr = np.array([
    [2, 3, 5],
    [14, 6, 8],
    [11, 13, 17],
    [7, 10, 13]
])
```

```python
def prime_rows(arr):
    result = []

    for row in arr:
        for num in row:
            if num > 1:
                is_prime = True

                for i in range(2, num):
                    if num % i == 0:
                        is_prime = False
                        break

                if is_prime:
                    result.append(row)
                    break

    return np.array(result)

print(prime_rows(arr))
```

# part 5

```python
def checkerboard():
    board = np.zeros((8, 8))

    board[::2, ::2] = 1
    board[1::2, 1::2] = 1

    return board

print(checkerboard())
```

```python
def reverse_checkerboard():
    board = np.zeros((8, 8))

    board[::2, 1::2] = 1
    board[1::2, ::2] = 1

    return board

print(reverse_checkerboard())
```

# part 6

```python
def expansion(strings, n):
    return np.array([(" " * n).join(word) for word in strings])

universe = np.array(['galaxy', 'clusters', 'blackholes'])

print(expansion(universe, 1))
print(expansion(universe, 2))
```

```python
def secondDimmest(stars):
    sorted_stars = np.sort(stars, axis=1)
    return sorted_stars[:, -2]
stars = np.array([
    [1123, 1456, 1789, 1324, 1876],
    [1567, 1987, 1678, 1405, 1589],
    [1345, 1654, 1523, 1109, 1923],
    [1298, 1890, 1367, 1784, 1432],
    [1823, 1756, 1489, 1672, 1550]
])
print(secondDimmest(stars))
```
