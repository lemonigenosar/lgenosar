```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = np.loadtxt("../../course_assignments/homework7/star_data.txt",
skiprows=1)
```

```python
data.shape
```

2642 rows x 2 columns

```python
magnitude = data[:, 0]
temperature = data[:, 1]

min_mag_index = np.argmin(magnitude)
max_mag_index = np.argmax(magnitude)

min_temp_index = np.argmin(temperature)
max_temp_index = np.argmax(temperature)

print("Min magnitude:",np.min(magnitude), "at index", [min_mag_index])
print("Max magnitude:",np.max(magnitude), "at index", [max_mag_index])


print("Min temperature:",np.min(temperature), "at index", [min_temp_index])
print("Max temperature:",np.max(temperature), "at index", [max_temp_index])
```

The star with magnitude 2.24 at index [np.int64(1988)] is brighter than the star with magnitude  12.49 at index [np.int64(264)] 

```python
log_magnitude = np.log(magnitude)
log_temperature = np.log(temperature)

brightest = np.argmin(magnitude)
dimmest = np.argmax(magnitude)
hottest = np.argmax(temperature)
coldest = np.argmin(temperature)

plt.figure(figsize=(8, 8))

scatter = plt.scatter(
    log_temperature,
    log_magnitude,
    c=temperature,
    cmap='coolwarm',
    label='Stars',
    s=2
)

plt.scatter(
    log_temperature[brightest],
    log_magnitude[brightest],
    color='black',
    marker='*',
    s=300,
    label='Brightest'
)

plt.scatter(
    log_temperature[dimmest],
    log_magnitude[dimmest],
    color='purple',
    marker='*',
    s=300,
    label='Dimmest'
)

plt.scatter(
    log_temperature[hottest],
    log_magnitude[hottest],
    color='green',
    marker='*',
    s=300,
    label='Hottest'
)

plt.scatter(
    log_temperature[coldest],
    log_magnitude[coldest],
    color='orange',
    marker='*',
    s=300,
    label='Coldest'
)

plt.gca().invert_xaxis()
plt.gca().invert_yaxis()

plt.xlabel('Temperature (log K)')
plt.ylabel('Magnitude')
plt.title('Hertzsprung-Russel Diagram', fontsize = 18)
plt.colorbar(scatter, label='Temperature (K)')
plt.legend()

plt.show()
```
